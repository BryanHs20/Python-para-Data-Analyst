# Analisis adicionales

## ¿Qué códigos postales tienen las casas más caras?

```python
analisis1 = (
    housing
    .groupby("zipcode")
    .agg(
        Precio_Promedio=("price","mean"),
        Numero_Casas=("id","count")
    )
    .sort_values("Precio_Promedio", ascending=False)
)

analisis1.head(10)
```

## ¿Qué tipo de vivienda ofrece mayor precio por metro cuadrado?

```python
housing["precio_m2"] = housing["price"] / housing["m2"]

analisis2 = (
    housing
    .groupby("bedrooms")
    .agg(
        Precio_m2=("precio_m2","mean"),
        Cantidad=("id","count")
    )
    .sort_values("Precio_m2", ascending=False)
)

analisis2
```
