## Query

```python
filter={
    'year': {
        '$gte': 2010, 
        '$lte': 2019
    }, 
    'imdb.rating': {
        '$gte': 8
    }
}

result = client['sample_mflix']['movies'].find(
  filter=filter
)
```

```python
import pprint
pp = pprint.PrettyPrinter(indent=4)

for doc in result:
    pp.pprint(doc)
```

```python
# Ejemplo 2: Buscar películas con "Dark" en el título (case insensitive)
filter_dark = {
    'title': {
        '$regex': 'Dark',
        '$options': 'i'
    }
}

result_dark = client['sample_mflix']['movies'].find(
    filter=filter_dark
)

for doc in result_dark:
    pp.pprint(doc)
```

```python
# Ejemplo 3 con operador $and - películas con rating >= 8 Y año entre 2010-2019
filter_and = {
    '$and': [
        {
            'year': {
                '$gte': 2010,
                '$lte': 2019
            }
        },
        {
            'imdb.rating': {
                '$gte': 8
            }
        }
    ]
}

result_and = client['sample_mflix']['movies'].find(
    filter=filter_and
)

print("Películas con rating >= 8 AND año 2010-2019:")
for doc in result_and:
    pp.pprint(doc)
```

```python
# Ejemplo 4 con operador $or - películas con rating >= 9 O año >= 2018
filter_or = {
    '$or': [
        {
            'imdb.rating': {
                '$gte': 9
            }
        },
        {
            'year': {
                '$gte': 2018
            }
        }
    ]
}

result_or = client['sample_mflix']['movies'].find(
    filter=filter_or
)

print("Películas con rating >= 9 OR año >= 2018:")
for doc in result_or:
    pp.pprint(doc)
```

## Añadir info

```python
# Ejemplo para añadir un nuevo campo a documentos en la base de datos
# Actualizar un documento específico agregando un nuevo campo

# Ejemplo 1: Agregar un nuevo campo a un documento específico
client['sample_mflix']['movies'].update_one(
    {'title': 'Very Semi-Serious'},  # Filtro para encontrar el documento
    {'$set': {'nuevo_campo': 'valor_ejemplo'}}  # Agregar nuevo campo
)
```

```python

# Ejemplo 2: Agregar un nuevo campo a múltiples documentos
# Agregar campo "visto" a todas las películas con rating >= 8
client['sample_mflix']['movies'].update_many(
    {'imdb.rating': {'$gte': 8}},
    {'$set': {'visto': False, 'fecha_agregado': '2024-01-01'}}
)
```

```python
# Ejemplo 3: Verificar la actualización
resultado_actualizado = client['sample_mflix']['movies'].find_one(
    {'title': 'Very Semi-Serious'}
)

print("Documento actualizado:")
pp.pprint(resultado_actualizado)
```

```python
# Ejemplo para actualizar datos existentes en documentos

# Ejemplo 1: Actualizar el campo 'visto' de una película específica
client['sample_mflix']['movies'].update_one(
    {'title': 'Very Semi-Serious'},
    {'$set': {'visto': True}}
)
```

```python
# Ejemplo para eliminar datos de la base de datos

# Ejemplo 1: Eliminar el campo 'nuevo_campo' de un documento específico
client['sample_mflix']['movies'].update_one(
    {'title': 'Very Semi-Serious'},
    {'$unset': {'nuevo_campo': ''}}
)

print("Campo 'nuevo_campo' eliminado del documento 'Very Semi-Serious'")
```
