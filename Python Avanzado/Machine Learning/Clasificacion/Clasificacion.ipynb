# Feature Engineering

```python
df2 = df.copy()
```

```python
# Eliminamos las columnas que no aportan informacion relevante
df2.drop(columns=['EmployeeCount', 'EmployeeID', 'Over18', 'StandardHours'], axis=1, inplace=True)
```

```python
# Cambio de tipo de dato a las columnas
df2['BusinessTravel'] = df2['BusinessTravel'].astype('category')
df2['Department'] = df2['Department'].astype('category')
df2['EducationField'] = df2['EducationField'].astype('category')
df2['Gender'] = df2['Gender'].astype('category')
```

```python
# Cambiamos la columna NumCompaniesWorked a entero
df2['NumCompaniesWorked'] = df2['NumCompaniesWorked'].fillna(0).astype(np.int64)
```

```python
df2.sample(5)
```

```python
# Obtenemos los distintos valores para ver si se hace un valanceo de clases
print(df2['EducationField'].value_counts())
print('---------------------')
print(df2['BusinessTravel'].value_counts())
print('---------------------')
print(df2['Department'].value_counts())
```

```python
# Forma manuela de asignar valores binarios
df2['Attrition'] = df2['Attrition'].replace({'Yes':1, 'No':0}, regex=True)
df2['Gender'] = df2['Gender'].replace({'Male':1, 'Female':0}, regex=True)
df2['EducationField'] = df2['EducationField'].replace({'Yes':1, 'No':0}, regex=True)
df2['BusinessTravel'] = df2['BusinessTravel'].replace({'Yes':1, 'No':0}, regex=True)
df2['Department'] = df2['Department'].replace({'Yes':1, 'No':0}, regex=True)
```

```python
# Forma automatica
df2.Attrition = df2.Attrition.astype('category').cat.codes
df2.Gender = df2.Gender.astype('category').cat.codes
df2.BusinessTravel = df2.BusinessTravel.astype('category').cat.codes
df2.Department = df2.Department.astype('category').cat.codes
df2.EducationField = df2.EducationField.astype('category').cat.codes
df2.MaritalStatus = df2.MaritalStatus.astype('category').cat.codes
df2.JobRole = df2.JobRole.astype('category').cat.codes

# Usaremos este
# Es mas escalable que el metodo anterior
```

```python
df2.sample(3)
```

```python
# Estandarizaremos
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df2['MonthlyIncome_scale'] = StandardScaler().fit_transform(df2[['MonthlyIncome']])
```

```python
df2.drop(columns='MonthlyIncome',axis=1, inplace=True)
```

```python
cols_num = df2.columns[[np.issubdtype(dt, np.number) for dt in df2.dtypes]]
```

# Data Split

```python
from sklearn.model_selection import train_test_split
x = df2.drop('Attrition', axis=1)
y = df2['Attrition']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=1)
```

```python
x.head(5)
```

---

# Regresion Logistica

```python
print(df2['Attrition'].value_counts())
print(df2['Attrition'].value_counts(normalize=True))
```

```python
from sklearn import metrics
from sklearn.linear_model import LogisticRegression

# Adaptamos el modelo a los x_train y y_train
model = LogisticRegression(solver='liblinear', class_weight='balanced')

model.fit(x_train, y_train)

# Hace la prediccion en x_test
y_pred = model.predict(x_test)
```

```python
y_pred[0:200]
```

```python
# Construimos un dataset con los valores reales y predichos
z = x_test.copy()
z['Attrition_Real'] = y_test
z['Attrition_Prediccion'] = y_pred
z[['Attrition_Real', 'Attrition_Prediccion']].head(10)
```

```python
# Generamos los coeficientes
coef_df = pd.DataFrame(model.coef_)
coef_df.columns = x.columns
# Incluye el interceptor
coef_df['intercept'] = model.intercept_
coef_df.round(2).T
```

# Matriz

```python
cm = metrics.confusion_matrix(y_test, y_pred, labels=[1,0])

df_cm = pd.DataFrame(cm, index = [i for i in ['1', '0']], columns= [ i for i in ['Prediccion 1', 'Prediccion 0']])
plt.figure(figsize=(6,4))
sns.heatmap(df_cm, annot=True, cmap='Accent', fmt='d').set_title
```

```python
TP = df_cm.iloc[0,0]
FP = df_cm.iloc[0,1]
FN = df_cm.iloc[1,0]
TN = df_cm.iloc[1,1]
```

```python
# Medidas del modelo
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, confusion_matrix

y_pred = model.predict(x_test)
model_score = accuracy_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print('Medidas del modelo')
print('------------------')
print('Accuracy:', round(model_score, 2))
print('Recall:', round(recall, 2))
print('Precision:', round(precision, 2))
print('F1 Score:', round(f1, 2))
```
