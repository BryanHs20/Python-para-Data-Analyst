## Serie de tiempo

```python
from sklearn.linear_model import LinearRegression
from datetime import datetime

import matplotlib.dates as mdates

# Preparar datos para la predicción
# Convertir AÑO_MES a formato datetime
df_long['FECHA'] = pd.to_datetime(df_long['AÑO'].astype(str) + '-' + 
                                   df_long['MES'].map({
                                       'ENERO': '01', 'FEBRERO': '02', 'MARZO': '03',
                                       'ABRIL': '04', 'MAYO': '05', 'JUNIO': '06',
                                       'JULIO': '07', 'AGOSTO': '08', 'SEPTIEMBRE': '09',
                                       'OCTUBRE': '10', 'NOVIEMBRE': '11', 'DICIEMBRE': '12'
                                   }) + '-01')

# Agregar delitos por fecha
delitos_por_fecha = df_long.groupby('FECHA')['DELITOS'].sum().reset_index().sort_values('FECHA')
```

```python
# Preparar datos para el modelo
X = np.arange(len(delitos_por_fecha)).reshape(-1, 1)
y = delitos_por_fecha['DELITOS'].values
```

```python
# Entrenar modelo de regresión lineal
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
# Generar predicciones para 2022
fecha_inicio_2022 = pd.to_datetime('2022-01-01')
fecha_fin_2022 = pd.to_datetime('2022-12-31')
fechas_2022 = pd.date_range(start=fecha_inicio_2022, end=fecha_fin_2022, freq='MS')

X_2022 = np.arange(len(delitos_por_fecha), len(delitos_por_fecha) + len(fechas_2022)).reshape(-1, 1)
predicciones_2022 = modelo.predict(X_2022)
```

## Clustering

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Preparación de datos para 2021
df_2021 = df_long[df_long['AÑO'] == 2021].copy()

peligrosidad_2021 = (
    df_2021
    .groupby('ENTIDAD')
    .agg(
        total_delitos_2021=('DELITOS', 'sum'),
        promedio_mensual_2021=('DELITOS', 'mean'),
        desv_std_mensual_2021=('DELITOS', 'std'),
        meses_reportados_2021=('MES', 'nunique')
    )
    .reset_index()
)

peligrosidad_2021['desv_std_mensual_2021'] = peligrosidad_2021['desv_std_mensual_2021'].fillna(0)
peligrosidad_2021['tasa_delitos_por_mes'] = (
    peligrosidad_2021['total_delitos_2021'] /
    peligrosidad_2021['meses_reportados_2021']
)
```

```python
# Normalización de las variables
features = ['total_delitos_2021', 'promedio_mensual_2021', 'desv_std_mensual_2021']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(peligrosidad_2021[features])
```

```python

# Búsqueda del número óptimo de clusters
k_range = range(2, 7)
inertia = []
silhouette = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertia.append(km.inertia_)
    silhouette.append(silhouette_score(X_scaled, labels))

best_k = k_range[int(np.argmax(silhouette))]
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
peligrosidad_2021['cluster'] = kmeans.fit_predict(X_scaled)
peligrosidad_2021['cluster'] = peligrosidad_2021['cluster'] + 1
```

```python
# Etiquetas de peligrosidad según el promedio de delitos por cluster
cluster_summary = (
    peligrosidad_2021
    .groupby('cluster')
    .agg(
        estados=('ENTIDAD', 'count'),
        promedio_total_delitos=('total_delitos_2021', 'mean'),
        promedio_mensual=('promedio_mensual_2021', 'mean'),
        promedio_std=('desv_std_mensual_2021', 'mean')
    )
    .reset_index()
    .sort_values('promedio_total_delitos', ascending=False)
    .reset_index(drop=True)
)

niveles = ['Alta', 'Media', 'Baja'][:len(cluster_summary)]
cluster_summary['nivel_peligrosidad'] = niveles
nivel_map = cluster_summary.set_index('cluster')['nivel_peligrosidad'].to_dict()
peligrosidad_2021['nivel_peligrosidad'] = peligrosidad_2021['cluster'].map(nivel_map)

# Resultados
print("Número de clusters seleccionado:", best_k)
print("\nResumen de clusters por peligrosidad:")
print(cluster_summary.to_string(index=False))

print("\nPrimeras filas del dataframe final de peligrosidad:")
print(
    peligrosidad_2021
    .sort_values('total_delitos_2021', ascending=False)
    .loc[:, [
        'ENTIDAD', 'total_delitos_2021', 'promedio_mensual_2021',
        'desv_std_mensual_2021', 'cluster', 'nivel_peligrosidad'
    ]]
    .head(15)
    .to_string(index=False)
)
```

```python
# Gráfica de codo y silueta
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(list(k_range), inertia, marker='o', color='#1f77b4')
axes[0].set_title('Elbow: Inercia vs Número de clusters')
axes[0].set_xlabel('Número de clusters')
axes[0].set_ylabel('Inercia')

axes[1].plot(list(k_range), silhouette, marker='o', color='#ff7f0e')
axes[1].set_title('Silhouette Score vs Número de clusters')
axes[1].set_xlabel('Número de clusters')
axes[1].set_ylabel('Silhouette Score')

plt.tight_layout()
plt.show()
```

### (a)  Número promedio de delitos agrupados por municipio y año-mes de ocurrencia de estos.

```python
query_a = """
SELECT 
    ENTIDAD AS municipio,
    AÑO_MES,
    AVG(DELITOS) AS promedio_delitos
FROM Fuero_federal
GROUP BY ENTIDAD, AÑO_MES
ORDER BY ENTIDAD, AÑO_MES;
"""

df_a = pd.read_sql(query_a, conn)
df_a.head(10)

```

### (b)  El delito más común por municipio

```python
query_b = """
WITH ConteoDelitos AS (
    SELECT 
        ENTIDAD,
        TIPO AS delito,
        SUM(DELITOS) AS total_delitos,
        ROW_NUMBER() OVER (
            PARTITION BY ENTIDAD 
            ORDER BY SUM(DELITOS) DESC
        ) AS rn
    FROM Fuero_federal
    GROUP BY ENTIDAD, TIPO
)
SELECT 
    ENTIDAD,
    delito,
    total_delitos
FROM ConteoDelitos
WHERE rn = 1
ORDER BY ENTIDAD;
"""

df_b = pd.read_sql(query_b, conn)
df_b.head(10)
```

### (c)  La temporada con más crímenes en México

```python
query_c = """
SELECT 
    CASE 
        WHEN MES IN ('MARZO','ABRIL','MAYO') THEN 'PRIMAVERA'
        WHEN MES IN ('JUNIO','JULIO','AGOSTO') THEN 'VERANO'
        WHEN MES IN ('SEPTIEMBRE','OCTUBRE','NOVIEMBRE') THEN 'OTOÑO'
        WHEN MES IN ('DICIEMBRE','ENERO','FEBRERO') THEN 'INVIERNO'
    END AS temporada,
    SUM(DELITOS) AS total_delitos
FROM Fuero_federal
GROUP BY 
    CASE 
        WHEN MES IN ('MARZO','ABRIL','MAYO') THEN 'PRIMAVERA'
        WHEN MES IN ('JUNIO','JULIO','AGOSTO') THEN 'VERANO'
        WHEN MES IN ('SEPTIEMBRE','OCTUBRE','NOVIEMBRE') THEN 'OTOÑO'
        WHEN MES IN ('DICIEMBRE','ENERO','FEBRERO') THEN 'INVIERNO'
    END
ORDER BY total_delitos DESC;
"""

df_c = pd.read_sql(query_c, conn)
df_c
```

###  (d)  El número promedio de delitos agrupados por tipo de delito y año y mes en la ciudad de México ordenados de mayor a menor

```python
query_d = """
SELECT 
    TIPO,
    AÑO,
    MES,
    AVG(DELITOS) AS promedio_delitos
FROM Fuero_federal
WHERE ENTIDAD = 'CIUDAD DE MEXICO'
GROUP BY TIPO, AÑO, MES
ORDER BY promedio_delitos DESC;
"""

df_d = pd.read_sql(query_d, conn)
df_d.head(10)
```
