## Masajeo de datos

## Analisis

```python
# Estadistica descriptiva de la columna unit price
unit_price_mean = df['Unit price'].mean()
unit_price_median = df['Unit price'].median()
unit_price_moda = df['Unit price'].mode()

print('Estadística descriptiva de la columna Precio Unitario:')
print(f'La media del Precio Unitario es de: {unit_price_mean:.2f}')
print(f'La mediana del Precio Unitario es de: {unit_price_median:.2f}')
print(f'La moda del Precio Unitario es de: {unit_price_moda[0]:.2f}')
```

```python
# Estadistica descriptiva de la columna quantity
quantity_mean = df['Quantity'].mean()
quantity_median = df['Quantity'].median()
quantity_moda = df['Quantity'].mode()

print('Estadística descriptiva de la columna Cantidad:')
print(f'La media de la Cantidad es: {quantity_mean:.2f}')
print(f'La mediana de la Cantidad es: {quantity_median:.2f}')
print(f'La moda de la Cantidad es: {quantity_moda[0]:.2f}')
```

## Obtencion del ticket promedio
1) por ciudad,
2) por product line y ciudad,
3) por género,
4) por género y product line

```python
city_ticket_avg = df.groupby('City')['Total'].mean().round(2)
print(f"Ticket promedio por ciudad: {city_ticket_avg}")
```

```python
pdt_line_city_ticket_avg = df.groupby(['City', 'Product line'])['Total'].mean().round(2)
print(f'Ticket promedio por linea de producciopn y ciudad: {pdt_line_city_ticket_avg}')
```

```python
gender_ticket_avg = df.groupby('Gender')['Total'].mean().round(2)
print(f'Ticket promedio por género: {gender_ticket_avg}')
```

```python
pdt_line_gender_ticket_avg = df.groupby(['Gender', 'Product line'])['Total'].mean().round(2)
print(f'Ticket promedio por linea de producciopn y género: {pdt_line_gender_ticket_avg}')
```

## Usar el método rank para generar top 5 de:
1) ventas por ciudad,
2) ventas por member,
3) ventas por payment. Obtener además el % de aporte de cada categoría.  (P.ej. La ciudad X aporta con el 25% del total de ventas, etc.)

```python
# Top ventas por ciudad
city_sales = df.groupby('City')['Total'].sum().sort_values(ascending=False)
city_sales_pct = (city_sales / city_sales.sum() * 100).round(2)
top_city = pd.DataFrame({'Total Ventas': city_sales, 'Porcentaje': city_sales_pct})
print("Top ventas por ciudad:")
print(top_city)
print("\n")
```

```python
# Top ventas por member (Customer type)
member_sales = df.groupby('Customer type')['Total'].sum().sort_values(ascending=False)
member_sales_pct = (member_sales / member_sales.sum() * 100).round(2)
top_member = pd.DataFrame({'Total Ventas': member_sales, 'Porcentaje': member_sales_pct})
print("Top ventas por customer type:")
print(top_member)
print("\n")
```

```python
# Top ventas por payment
payment_sales = df.groupby('Payment')['Total'].sum().sort_values(ascending=False)
payment_sales_pct = (payment_sales / payment_sales.sum() * 100).round(2)
top_payment = pd.DataFrame({'Total Ventas': payment_sales, 'Porcentaje': payment_sales_pct})
print("Top ventas por método de pago:")
print(top_payment)
```

```python
# Top ventas por genero
gender_sales = df.groupby('Gender')['Total'].sum().sort_values(ascending=False)
gender_sales_pct = (gender_sales / gender_sales.sum() * 100).round(2)
top_gender = pd.DataFrame({'Total Ventas': gender_sales, 'Porcentaje': gender_sales_pct})
print("Top ventas por genero:")
print(top_gender)
```

## Matriz Correlacional
Cálculo de una matriz de correlación entre la hora (sin minutos) y el total, y otra que correlacione el unit_price con el rating de la transacción, para validar si los productos más caros son los que dejan más margen

```python
# Extraer la hora sin minutos
df['Hour'] = df['Time'].str.split(':').str[0].astype(int)

# Matriz de correlación entre hora y total
correlation_hour_total = df[['Hour', 'Total']].corr().round(4)
print("Correlación entre Hora y Total de ventas:")
print(correlation_hour_total)
```

```python
# Matriz de correlación entre unit_price y rating
correlation_price_rating = df[['Unit price', 'Rating']].corr().round(4)
print("Correlación entre Precio Unitario y Rating:")
print(correlation_price_rating)
```

```python
# Visualizar las correlaciones
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Heatmap correlación hora-total
sns.heatmap(correlation_hour_total, annot=True, cmap='coolwarm', center=0, ax=axes[0], vmin=-1, vmax=1)
axes[0].set_title('Correlación: Hora vs Total de Ventas')

# Heatmap correlación price-rating
sns.heatmap(correlation_price_rating, annot=True, cmap='coolwarm', center=0, ax=axes[1], vmin=-1, vmax=1)
axes[1].set_title('Correlación: Precio Unitario vs Rating')

plt.tight_layout()
plt.show()
```
