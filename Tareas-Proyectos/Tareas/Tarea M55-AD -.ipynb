## Se realizara un cambio de tipo de dato a las siguientes columnas:
- Date 
- Bathrooms
- Floors

Ya que el tipo de dato es float, cosa que no debe de ser asi, ya que a como se maneja la imformacion, deben de ser el tipo entero

En caso de Date, este debe de cambiarse a un mejor formato, ejem: 'DD/MM/YYYY', ya que a la hora de trabajar con looker studio y querer manejar graficos con fechas, este solo acepta el formato antes mencionado 

```python
# Convertimos las columnas bathrooms y floors a tipo entero
df['bathrooms'] = df['bathrooms'].astype('int64')
df['floors'] = df['floors'].astype('int64')

df.dtypes
```

```python
# Convertimos la columna a formato tipo datetime y con la serie DD/MM/YYYY
df['date'] = pd.to_datetime(df['date'], format='%Y%m%dT%H%M%S').dt.strftime('%d/%m/%Y')

df['date'].head()
```

## Comparacion de analisis obtenido

```python
# Agrupar por zipcode y año de construcción, calcular métricas y ordenar por precio promedio descendente
analysis = df.groupby(['zipcode', 'yr_built']).agg({
    'price': 'mean',
    'bedrooms': 'mean',
    'bathrooms': 'mean'
}).reset_index()

analysis.columns = ['zipcode', 'yr_built', 'precio_promedio', 'bedrooms_promedio', 'bathrooms_promedio']
analysis = analysis.sort_values('precio_promedio', ascending=False).round(2).head(50)

analysis
```

```python
analysis_zipcode_only = df.groupby(['zipcode']).agg({
    'price': 'mean',
    'bedrooms': 'mean',
    'bathrooms': 'mean'
}).reset_index()

analysis_zipcode_only.columns = ['zipcode', 'precio_promedio', 'bedrooms_promedio', 'bathrooms_promedio']
analysis_zipcode_only = analysis_zipcode_only.sort_values('precio_promedio', ascending=False).round(2).head(50)

analysis_zipcode_only
```

```python
# Convertir square feet a metros cuadrados (1 sq ft = 0.092903 m²)
# Calcular precio por m² agrupado por zipcode
precio_por_m2 = df.groupby('zipcode').apply(
    lambda x: (x['price'].sum() / (x['sqft_living'].sum() * 0.092903)).round(2)
).reset_index()

precio_por_m2.columns = ['zipcode', 'precio_por_m2']
precio_por_m2 = precio_por_m2.sort_values('precio_por_m2', ascending=False)

precio_por_m2
```
