### (a) Número promedio de delitos agrupados por municipio y año-mes de ocurrencia

```python
# (a) Número promedio de delitos agrupados por municipio y año-mes de ocurrencia
# Primero, reorganizamos los datos para tener una estructura año-mes

months = ['ENERO', 'FEBRERO', 'MARZO', 'ABRIL', 'MAYO', 'JUNIO', 
          'JULIO', 'AGOSTO', 'SEPTIEMBRE', 'OCTUBRE', 'NOVIEMBRE', 'DICIEMBRE']

# Convertir de formato ancho a largo
df_long = df.melt(id_vars=['AÑO', 'ENTIDAD', 'LEY', 'CONCEPTO', 'TIPO', 'INEGI'],
                   value_vars=months,
                   var_name='MES',
                   value_name='DELITOS')

# Crear columna AÑO-MES
df_long['AÑO_MES'] = df_long['AÑO'].astype(str) + '-' + df_long['MES']

# Agrupar por ENTIDAD (municipio) y AÑO_MES, calcular promedio
promedio_delitos = df_long.groupby(['ENTIDAD', 'AÑO_MES'])['DELITOS'].sum().reset_index().round(2)
promedio_delitos.columns = ['MUNICIPIO', 'AÑO_MES', 'TOTAL_DELITOS']

# Calcular promedio por municipio
promedio_por_municipio = promedio_delitos.groupby('MUNICIPIO')['TOTAL_DELITOS'].mean().reset_index().round(2)
promedio_por_municipio.columns = ['MUNICIPIO', 'PROMEDIO_DELITOS']
promedio_por_municipio = promedio_por_municipio.sort_values('PROMEDIO_DELITOS', ascending=False)

promedio_por_municipio.head(15)
```

### (a)  Número promedio de delitos agrupados por municipio y año-mes de ocurrencia de estos.

```python
query_a = """
SELECT 
    ENTIDAD AS municipio,
    AÑO_MES,
    AVG(DELITOS) AS promedio_delitos
FROM Fuero_federal
GROUP BY ENTIDAD, AÑO_MES
ORDER BY ENTIDAD, AÑO_MES;
"""

df_a = pd.read_sql(query_a, conn)
df_a.head(10)

```

### (b)  El delito más común por municipio

```python
query_b = """
WITH ConteoDelitos AS (
    SELECT 
        ENTIDAD,
        TIPO AS delito,
        SUM(DELITOS) AS total_delitos,
        ROW_NUMBER() OVER (
            PARTITION BY ENTIDAD 
            ORDER BY SUM(DELITOS) DESC
        ) AS rn
    FROM Fuero_federal
    GROUP BY ENTIDAD, TIPO
)
SELECT 
    ENTIDAD,
    delito,
    total_delitos
FROM ConteoDelitos
WHERE rn = 1
ORDER BY ENTIDAD;
"""

df_b = pd.read_sql(query_b, conn)
df_b.head(10)
```

### (c)  La temporada con más crímenes en México

```python
query_c = """
SELECT 
    CASE 
        WHEN MES IN ('MARZO','ABRIL','MAYO') THEN 'PRIMAVERA'
        WHEN MES IN ('JUNIO','JULIO','AGOSTO') THEN 'VERANO'
        WHEN MES IN ('SEPTIEMBRE','OCTUBRE','NOVIEMBRE') THEN 'OTOÑO'
        WHEN MES IN ('DICIEMBRE','ENERO','FEBRERO') THEN 'INVIERNO'
    END AS temporada,
    SUM(DELITOS) AS total_delitos
FROM Fuero_federal
GROUP BY 
    CASE 
        WHEN MES IN ('MARZO','ABRIL','MAYO') THEN 'PRIMAVERA'
        WHEN MES IN ('JUNIO','JULIO','AGOSTO') THEN 'VERANO'
        WHEN MES IN ('SEPTIEMBRE','OCTUBRE','NOVIEMBRE') THEN 'OTOÑO'
        WHEN MES IN ('DICIEMBRE','ENERO','FEBRERO') THEN 'INVIERNO'
    END
ORDER BY total_delitos DESC;
"""

df_c = pd.read_sql(query_c, conn)
df_c
```

###  El número promedio de delitos agrupados por tipo de delito y año y mes en la ciudad de México ordenados de mayor a menor

```python
query_d = """
SELECT 
    TIPO,
    AÑO,
    MES,
    AVG(DELITOS) AS promedio_delitos
FROM Fuero_federal
WHERE ENTIDAD = 'CIUDAD DE MEXICO'
GROUP BY TIPO, AÑO, MES
ORDER BY promedio_delitos DESC;
"""

df_d = pd.read_sql(query_d, conn)
df_d.head(10)
```
